```python
train_df = train_df.drop('State', axis=1)
test_df = test_df.drop('State', axis=1)
```

```python
train_df['International plan'] = train_df['International plan'].map({'Yes':1,'No':0})
train_df['Voice mail plan'] = train_df['Voice mail plan'].map({'Yes':1,'No':0})

test_df['International plan'] = test_df['International plan'].map({'Yes':1,'No':0})
test_df['Voice mail plan'] = test_df['Voice mail plan'].map({'Yes':1,'No':0})
```

```python
train_df['Churn'] = train_df['Churn'].map({True:1, False:0})
test_df['Churn'] = test_df['Churn'].map({True:1, False:0})
```

```python
train_df['Total usage'] = (
    train_df['Total day minutes'] +
    train_df['Total eve minutes'] +
    train_df['Total night minutes']
)

test_df['Total usage'] = (
    test_df['Total day minutes'] +
    test_df['Total eve minutes'] +
    test_df['Total night minutes']
)
```

```python
train_df['Avg day call duration'] = train_df['Total day minutes'] / (train_df['Total day calls'] + 1)
test_df['Avg day call duration'] = test_df['Total day minutes'] / (test_df['Total day calls'] + 1)
```

```python
train_df['Charge per minute'] = train_df['Total day charge'] / (train_df['Total day minutes'] + 1)
test_df['Charge per minute'] = test_df['Total day charge'] / (test_df['Total day minutes'] + 1)
```

```python
train_df['Service call ratio'] = train_df['Customer service calls'] / (train_df['Account length'] + 1)
test_df['Service call ratio'] = test_df['Customer service calls'] / (test_df['Account length'] + 1)
```

```python
from imblearn.over_sampling import SMOTE
```

```python
X_train = train_df.drop('Churn', axis=1)
y_train = train_df['Churn']
```

```python
sm = SMOTE(random_state=42)

X_train_res, y_train_res = sm.fit_resample(X_train, y_train)
```

```python
import pandas as pd

pd.Series(y_train_res).value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_res)
```

```python
X_test = test_df.drop('Churn', axis=1)

X_test_scaled = scaler.transform(X_test)
```

```python
X_test = test_df.drop('Churn', axis=1)
y_test = test_df['Churn']
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
from xgboost import XGBClassifier
```

```python
xgb = XGBClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
```

```python
xgb.fit(X_train_scaled, y_train_res)
```

```python
y_pred_xgb = xgb.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred_xgb))

print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_xgb))

print("\nClassification Report:\n", classification_report(y_test, y_pred_xgb))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_xgb)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='coolwarm')

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix (XGBoost)")

plt.show()
```

```python
sns.heatmap(cm, annot=True, fmt='d', cmap='coolwarm',
            xticklabels=['No Churn','Churn'],
            yticklabels=['No Churn','Churn'])
```

```python
from sklearn.metrics import roc_auc_score

y_prob = xgb.predict_proba(X_test_scaled)[:,1]

print("ROC-AUC Score:", roc_auc_score(y_test, y_prob))
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [4, 6, 8],
    'learning_rate': [0.05, 0.1],
    'subsample': [0.8, 1.0]
}
```

```python
grid = GridSearchCV(
    estimator=XGBClassifier(random_state=42),
    param_grid=param_grid,
    cv=3,
    scoring='recall',   # important for churn
    n_jobs=-1,
    verbose=1
)
```

```python
grid.fit(X_train_scaled, y_train_res)
```

```python
best_model = grid.best_estimator_

print("Best Parameters:", grid.best_params_)
```

```python
y_pred_best = best_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(y_test, y_pred_best))
print("\nReport:\n", classification_report(y_test, y_pred_best))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt
```

```python
# Logistic Regression
y_prob_lr = lr.predict_proba(X_test_scaled)[:,1]

# Random Forest
y_prob_rf = rf.predict_proba(X_test_scaled)[:,1]

# XGBoost
y_prob_xgb = xgb.predict_proba(X_test_scaled)[:,1]
```

```python
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)
fpr_xgb, tpr_xgb, _ = roc_curve(y_test, y_prob_xgb)
```

```python
auc_lr = roc_auc_score(y_test, y_prob_lr)
auc_rf = roc_auc_score(y_test, y_prob_rf)
auc_xgb = roc_auc_score(y_test, y_prob_xgb)
```

```python
plt.figure(figsize=(8,6))

plt.plot(fpr_lr, tpr_lr, label=f"Logistic Regression (AUC = {auc_lr:.2f})")
plt.plot(fpr_rf, tpr_rf, label=f"Random Forest (AUC = {auc_rf:.2f})")
plt.plot(fpr_xgb, tpr_xgb, label=f"XGBoost (AUC = {auc_xgb:.2f})")

plt.plot([0,1], [0,1], linestyle='--')  # diagonal line

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")

plt.legend()
plt.show()
```
